# AssoMemBench — Quickstart
Runs fully offline with the mock backend (no API key). Flip `backend="openai"` for real models.

```python
import sys, os
sys.path.insert(0, os.path.abspath('../src'))
sys.path.insert(0, os.path.abspath('../eval'))
from assomem import AssociativeMemoryAgent, AgentConfig, BenchItem, LLM
```

```python
agent = AssociativeMemoryAgent(LLM(backend='mock'), AgentConfig())
agent.ingest([
    'User: I spend weekends hiking in the mountains.',
    'User: I just bought a mirrorless camera for landscapes.',
    'User: I love catching the sunrise.'])
print('stored notes:', len(agent.notes), '| assoc edges:', len(agent.edges))
```

```python
item = BenchItem(item_id='demo', source='syn', scenario='associative',
    stored_context=[], query='Suggest a weekend trip.',
    options=['A sunrise photography trek to a mountain ridge', 'An all-night nightclub tour'],
    gold='A sunrise photography trek to a mountain ridge')
ans = agent.answer(item)
print('chosen:', ans.answer)
print('confidence:', round(ans.confidence,3))
print('used evidence ids:', ans.used_evidence)
```
